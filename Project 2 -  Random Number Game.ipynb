```python
import random

top_of_range = input("Type a number: ")

if top_of_range.isdigit(): #returns true/false
    top_of_range = int(top_of_range)  #int(25) --> 25

    if top_of_range == 0:
        print("Please type a number larger than 0.")
        quit()
        
else:
    print("Please type a number.")
    quit()
    
random_number  = random.randint(0,top_of_range)
guesses = 0
while True:
    guesses+=1
    user_guess = input("Make a guess: ")

    if user_guess.isdigit():#returns true/false
        user_guess = int(user_guess)  #int(25) --> 25

    else:
        print("Please type a number.")
    
    if user_guess == random_number:
        print("You got it!")

        break
    elif user_guess > random_number:
        print("You were above the number")
    else:
        print("You were below the number")

print("You got it in", guesses, "guesses! Bonjour") #print statment convert str --> int
```
